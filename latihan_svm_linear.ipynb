## 1. Konsep Dasar & Dataset Studi Kasus
**Penjelasan (Note):**
- **Support Vector Machine (SVM)** bekerja dengan mencari garis atau bidang pemisah optimal (*hyperplane*) yang memaksimalkan jarak (*margin*) antara dua kelas data.
- **Support Vectors** adalah titik-titik data dari masing-masing kelas yang posisinya paling dekat dengan *hyperplane* dan menentukan kemiringan serta letak garis pemisah.
- Pada studi kasus ini digunakan 8 data mahasiswa dengan 2 fitur prediktor:
  - $X_1$: **Lama Belajar** (jam)
  - $X_2$: **Lama Tidur** (jam)
  - $Y$: **Status Kelulusan** (`Lulus` = $+1$ / `1`, `Tidak` = $-1$ / `0`)

## 2. Implementasi SVM dengan Python
### Tahap 1: Import Library
**Penjelasan (Note):**
- `import numpy as np`: Mengimpor library **NumPy** untuk operasi vektor, matriks, dan perhitungan aljabar linear.
- `import matplotlib.pyplot as plt`: Mengimpor **Matplotlib** untuk membuat grafik *scatter plot*, garis *hyperplane*, dan batas *margin*.
- `from sklearn.svm import SVC`: Mengimpor kelas **Support Vector Classifier** (`SVC`) dari library **Scikit-Learn** untuk melatih model klasifikasi SVM secara otomatis.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

### Tahap 2: Persiapan Data
**Penjelasan (Note):**
- `X1`: Array NumPy berisi fitur pertama yaitu **Lama Belajar**.
- `X2`: Array NumPy berisi fitur kedua yaitu **Lama Tidur**.
- `y_manual`: Label kelas dalam format matematis SVM (`1` untuk **Lulus**, `-1` untuk **Tidak Lulus**).
- `X = np.column_stack((X1, X2))`: Menggabungkan vektor `X1` dan `X2` menjadi matriks fitur 2D berukuran `(8, 2)` sesuai format standar input Scikit-Learn.
- `y`: Label biner (`1` untuk **Lulus**, `0` untuk **Tidak Lulus**) yang digunakan untuk pelatihan menggunakan `SVC`.

```python
# Dataset studi kasus: Lama Belajar (X1), Lama Tidur (X2), dan Status Lulus (y)
X1 = np.array([2, 3, 4, 5, 1, 2, 1, 2], dtype=float)
X2 = np.array([8, 7, 6, 7, 6, 5, 4, 3], dtype=float)

# Label versi matematis (+1 = Lulus, -1 = Tidak)
y_manual = np.array([1, 1, 1, 1, -1, -1, -1, -1], dtype=float)

# Menggabungkan fitur X1 dan X2 menjadi matriks 2 dimensi untuk Scikit-Learn
X = np.column_stack((X1, X2))
y = np.array([1, 1, 1, 1, 0, 0, 0, 0])

print("Matriks Fitur X:\n", X)
print("Label y:", y)
```

### Tahap 3A: Membuat dan Melatih Model SVM (Perhitungan Manual dengan NumPy)
**Penjelasan (Note):**
- Pada pendekatan manual menggunakan **NumPy**, kita menginisialisasi bobot $w = [w_1, w_2]$ dan bias $b$, lalu memperbaruinya secara iteratif menggunakan algoritma *Gradient Descent* (*Hinge Loss* + regularisasi).
- Kondisi klasifikasi benar dengan margin penuh terpenuhi jika $y_i (w \cdot x_i + b) \ge 1$.
- Jika data melanggar margin ($y_i (w \cdot x_i + b) < 1$), bobot diperbarui menggunakan gradien penalti ($C$); jika sudah benar, hanya dilakukan peluruhan bobot (*regularization*).

```python
# Implementasi Linear SVM dari nol (Manual dengan NumPy)
w = np.zeros(2)
b = 0.0
learning_rate = 0.01
C = 10.0
epochs = 2000

for epoch in range(epochs):
    for i in range(len(X)):
        condition = y_manual[i] * (np.dot(X[i], w) + b) >= 1
        if condition:
            w -= learning_rate * (2 * (1 / epochs) * w)
        else:
            w -= learning_rate * (2 * (1 / epochs) * w - C * y_manual[i] * X[i])
            b += learning_rate * (C * y_manual[i])

print(f"[Manual NumPy] Bobot w = {w}, Bias b = {b:.4f}")
```

### Tahap 3B: Membuat dan Melatih Model SVM (Menggunakan Scikit-Learn)
**Penjelasan (Note):**
- `clf = SVC(kernel='linear', C=10)`: Membuat objek model *Support Vector Classifier* dengan **kernel linear** (untuk pemisah garis lurus) dan parameter regularisasi `C=10`.
- `clf.fit(X, y)`: Melatih model SVM menggunakan matriks fitur `X` dan label `y` untuk menemukan *hyperplane* optimal secara eksak.
- `clf.coef_[0]` dan `clf.intercept_[0]`: Mengambil nilai bobot optimal ($w_1, w_2$) dan bias ($b$) hasil pelatihan.
- `clf.support_vectors_`: Menampilkan koordinat titik-titik data yang terpilih sebagai *support vectors*.
- `clf.predict(...)`: Melakukan prediksi kelas untuk data baru (misalnya mahasiswa dengan Lama Belajar = 3 jam dan Lama Tidur = 6 jam).

```python
# Membuat dan melatih model SVM menggunakan Scikit-Learn
clf = SVC(kernel='linear', C=10)
clf.fit(X, y)

# Mengambil parameter hyperplane (w dan b) serta support vectors
w_sk = clf.coef_[0]
b_sk = clf.intercept_[0]

print(f"[Scikit-Learn] Bobot w = {w_sk}, Bias b = {b_sk:.4f}")
print("Support Vectors:\n", clf.support_vectors_)

# Contoh prediksi data baru: Belajar 3 jam, Tidur 6 jam
data_baru = np.array([[3, 6]])
prediksi = clf.predict(data_baru)
print(f"Prediksi untuk (Belajar=3, Tidur=6): {'Lulus (1)' if prediksi[0] == 1 else 'Tidak (0)'}")
```

### Tahap 4: Visualisasi Decision Boundary dan Margin
**Penjelasan (Note):**
- Persamaan garis *hyperplane* adalah $w_1 x_1 + w_2 x_2 + b = 0$, sehingga nilai $x_2$ dapat dihitung sebagai:
  $$x_2 = -\frac{w_1 x_1 + b}{w_2}$$
- Garis **Upper Margin** ($+1$) dan **Lower Margin** ($-1$) digambar sejajar dengan *hyperplane* pada jarak $\pm \frac{1}{w_2}$ dari garis pemisah utama.
- Titik-titik yang dilingkari menandakan **Support Vectors** yang menjadi tumpuan pembatas kelas **Lulus** (biru) dan **Tidak Lulus** (merah).

```python
plt.figure(figsize=(8, 6))

# Plot titik data kelas Lulus (y = 1) dan Tidak (y = 0)
plt.scatter(X[y == 1, 0], X[y == 1, 1], color='blue', marker='x', s=80, label='Lulus (y = +1)')
plt.scatter(X[y == 0, 0], X[y == 0, 1], color='red', marker='x', s=80, label='Tidak (y = -1)')

# Menandai titik Support Vectors
plt.scatter(clf.support_vectors_[:, 0], clf.support_vectors_[:, 1],
            s=200, facecolors='none', edgecolors='black', linewidths=1.5, label='Support Vectors')

# Menghitung koordinat garis Hyperplane dan Margin
x_vals = np.linspace(0.5, 5.5, 100)
hyperplane = -(w_sk[0] * x_vals + b_sk) / w_sk[1]
margin_atas = -(w_sk[0] * x_vals + b_sk - 1) / w_sk[1]
margin_bawah = -(w_sk[0] * x_vals + b_sk + 1) / w_sk[1]

# Menggambar garis Hyperplane dan batas Margin
plt.plot(x_vals, hyperplane, 'g-', linewidth=2, label='Decision Boundary (Hyperplane)')
plt.plot(x_vals, margin_atas, 'k--', linewidth=1, label='Margin (+1 / -1)')
plt.plot(x_vals, margin_bawah, 'k--', linewidth=1)

plt.xlim(0.5, 5.5)
plt.ylim(2.5, 8.8)
plt.xlabel('Lama Belajar (X1)')
plt.ylabel('Lama Tidur (X2)')
plt.title('Visualisasi Linear SVM: Decision Boundary & Margin')
plt.legend(loc='lower left')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
