### 1. Mengimpor Library Pandas dan Membaca Dataset
**Penjelasan (Note):**
- `import pandas as pd`: Mengimpor library `pandas` (dengan alias `pd`) yang berfungsi untuk mengolah dan memanipulasi data berbentuk tabel (DataFrame).
- `df = pd.read_csv('diabetes.csv')`: Membaca file `diabetes.csv` dan menyimpannya ke dalam variabel `df` sebagai objek DataFrame.

```python
import pandas as pd

df = pd.read_csv('diabetes.csv')
```

### 2. Menampilkan 5 Baris Pertama DataFrame
**Penjelasan (Note):**
- `df.head()`: Menampilkan 5 baris teratas dari dataset `df` untuk melihat gambaran awal isi data serta nama-nama kolom yang tersedia.

```python
df.head()
```

### 3. Memeriksa Informasi Struktur dan Tipe Data Dataset
**Penjelasan (Note):**
- `df.info()`: Mengecek apakah terdapat nilai yang hilang (*missing values* / `null`) pada dataset serta memastikan seluruh kolom sudah bertipe numerik (`int64` dan `float64`) agar siap diproses oleh model machine learning.

```python
df.info()
```

### 4. Memisahkan Atribut (Fitur) dan Label (Target)
**Penjelasan (Note):**
- `X = df[df.columns[:8]]`: Mengambil 8 kolom pertama (`Pregnancies` sampai `Age`) sebagai atribut/fitur prediktor dan menyimpannya pada variabel `X`.
- `y = df['Outcome']`: Mengambil kolom `Outcome` (bernilai `0` untuk negatif diabetes dan `1` untuk positif diabetes) sebagai label target dan menyimpannya pada variabel `y`.

```python
# memisahkan atribut pada dataset dan menyimpannya pada sebuah variabel
X = df[df.columns[:8]]

# memisahkan label pada dataset dan menyimpannya sebagai sebuah variabel
y = df['Outcome']
```

### 5. Melakukan Standardisasi Nilai Fitur
**Penjelasan (Note):**
- `from sklearn.preprocessing import StandardScaler`: Mengimpor kelas `StandardScaler` dari modul `preprocessing` Scikit-Learn.
- `scaler = StandardScaler()`: Membuat objek `scaler` untuk menyamakan skala distribusi nilai antar fitur (mengubah rata-rata menjadi 0 dan standar deviasi menjadi 1).
- `scaler.fit(X)`: Menghitung nilai rata-rata (*mean*) dan standar deviasi dari setiap kolom fitur pada `X`.
- `X = scaler.transform(X)`: Menerapkan hasil perhitungan standardisasi ke data `X` sehingga fitur berskala besar (seperti `Insulin`) tidak mendominasi fitur berskala kecil (seperti `DiabetesPedigreeFunction`).

```python
from sklearn.preprocessing import StandardScaler

# standarisasi nilai-nilai dari dataset
scaler = StandardScaler()
scaler.fit(X)
X = scaler.transform(X)
```

### 6. Membagi Dataset Menjadi Data Latih (Training) dan Data Uji (Testing)
**Penjelasan (Note):**
- `from sklearn.model_selection import train_test_split`: Mengimpor fungsi `train_test_split` untuk membagi dataset.
- `X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)`: Memisahkan data menjadi **67% data latih** (`X_train`, `y_train`) dan **33% data uji** (`X_test`, `y_test`). Parameter `random_state=42` menjaga agar hasil pengacakan pembagian data selalu konsisten setiap kali kode dijalankan.

```python
from sklearn.model_selection import train_test_split

# memisahkan data untuk training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

### 7. Membuat dan Melatih Model Support Vector Classifier (SVC)
**Penjelasan (Note):**
- `from sklearn.svm import SVC`: Mengimpor algoritma *Support Vector Classifier* (`SVC`) dari modul `svm` Scikit-Learn.
- `clf = SVC()`: Menginisialisasi objek model klasifikasi SVM dengan konfigurasi parameter bawaan (*default*).
- `clf.fit(X_train, y_train)`: Melatih model `clf` untuk mempelajari pola hubungan antara fitur `X_train` dan label `y_train`.

```python
from sklearn.svm import SVC

# membuat objek SVC dan memanggil fungsi fit untuk melatih model
clf = SVC()
clf.fit(X_train, y_train)
```

### 8. Mengevaluasi Akurasi Model
**Penjelasan (Note):**
- `clf.score(X_test, y_test)`: Menguji performa model yang telah dilatih menggunakan data uji (`X_test`) dan membandingkan hasil prediksinya dengan label asli (`y_test`). Output akurasi yang dihasilkan adalah sekitar **0.7559** (75,59%).

```python
# Menampilkan skor akurasi prediksi
clf.score(X_test, y_test)
```
